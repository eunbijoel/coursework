###Approach 1: Traverse array elements and compare every element with the current largest element

Time complexity= 0(N)

```python
def App_1(array, n):
  #Initialize the largest element with the first element from the random array
  largest_element= array[0]
  #Traverse array elements from second and compare every element with the current largest element
  for i in range(1,n):
    if array[i]> largest_element:
      largest_element= array[i]
  #Return the largest element
  return largest_element
```

## Using for Loop

```python
# Define the two approaches to compare

#Approach1: Traverse array elements and compare every element with the current largest element
#Time complexity= O(N)
def App_1(array, n):
  #Initialize the largest element with the first element from the random array
  largest_element= array[0]
  #Traverse array elements from second and compare every element with the current largest element
  for i in range(1,n):
    if array[i]> largest_element:
      largest_element= array[i]
  #Return the largest element
  return largest_element


#Approach2:Sort array elements and find the last element of the array
#Time complexity= O(NlogN)
def App_2(array, n):
  #Sort the array_built in python library
  array.sort()
  #Return the last element of the array
  return array[n-1]

# define the input sizes to test
n = [10, 100, 1000]

# measure the elapsed time for each input size and approach
times1 = []
times2 = []

for size in n:
  array=np.random.randint(0, 100, size)
  
  start_time = time.time()
  App_1(array, size)
  end_time = time.time()
  times1.append(end_time - start_time)
    
  start_time = time.time()
  App_2(array, size)
  end_time = time.time()
  times2.append(end_time - start_time)
```

```python
# plot the results
plt.plot(n, times1, label='Approach 1')
plt.plot(n, times2, label='Approach 2')
plt.xlabel('Array Size (n)')
plt.ylabel('Elapsed Time')
plt.legend()
plt.show()
```
