#### 밸류팩터

먼저 밸류 지표에 해당하는 이익수익률을 계산해보겠습니다. 이익수익률은 이자 및 법인세 차감전이익(EBIT)을 기업가치(시가총액 + 순차입금)로 나눈 값입니다. 이를 분해하면 다음과 같습니다.

\begin{equation*}
\begin{split}
이익수익률 & = \frac{이자\,및\,법인세\,차감전이익}{기업\,가치} \\
& = \frac{이자\,및\,법인세\,차감전이익}{시가총액 + 순차입금} \\
& = \frac{당기순이익 + 법인세 + 이자비용}{시가총액 + 총부채 - 여유자금} \\
& = \frac{당기순이익 + 법인세 + 이자비용}{시가총액 + 총부채 - (현금 - max(0, 유동부채 - 유동자산 + 현금))}
\end{split}
\end{equation*}

**밸류 index**: PER, PBR, PSR, PCR

```python
# EPS and BPS for PBR and PER
cleaned_data['EPS'] = cleaned_data['지배주주순이익(천원)'] / cleaned_data['상장주식수(주)'] * 1000  # multiply by 1000 to convert 천원 to 원
cleaned_data['BPS'] = cleaned_data['지배주주순자산(자사주차감전)(천원)'] / cleaned_data['상장주식수(주)'] * 1000  # multiply by 1000 to convert 천원 to 원

# Calculate PBR
cleaned_data['PBR'] = cleaned_data['종가(원)'] / cleaned_data['BPS']

# Calculate PER
cleaned_data['PER'] = cleaned_data['종가(원)'] / cleaned_data['EPS']

# Calculate PCR
cleaned_data['CFO'] = cleaned_data['영업활동으로인한현금흐름(천원)'] / cleaned_data['상장주식수(주)'] * 1000  # Cash flow per share
cleaned_data['PCR'] = cleaned_data['종가(원)'] / cleaned_data['CFO']

# Calculate PSR
cleaned_data['SPS'] = cleaned_data['매출액(천원)'] / cleaned_data['상장주식수(주)'] * 1000  # Sales per share
cleaned_data['PSR'] = cleaned_data['종가(원)'] / cleaned_data['SPS']


cleaned_data[['PBR', 'PER', 'PCR','PSR']].head()
```

```python
#NA처리
columns_to_check = ['PER', 'PBR', 'PCR', 'PSR']

for column in columns_to_check:
    inf_rows_pos = cleaned_data[cleaned_data[column] == float('inf')]
    inf_rows_neg = cleaned_data[cleaned_data[column] == float('-inf')]

    if not inf_rows_pos.empty:
        print(f"Entries in {column} with 'inf':")
        print(inf_rows_pos[['year', 'Symbol', column]])

    if not inf_rows_neg.empty:
        print(f"Entries in {column} with '-inf':")
        print(inf_rows_neg[['year', 'Symbol', column]])

    # Replace infinities with NaN
    cleaned_data[column].replace([float('inf'), float('-inf')], np.nan, inplace=True)
```

```python
# Calculate the rank for each metric
cleaned_data['PBR_rank'] = cleaned_data.groupby('year')['PBR'].transform(lambda x: x.rank(ascending=True))
cleaned_data['PER_rank'] = cleaned_data.groupby('year')['PER'].transform(lambda x: x.rank(ascending=True))
cleaned_data['PCR_rank'] = cleaned_data.groupby('year')['PCR'].transform(lambda x: x.rank(ascending=True))
cleaned_data['PSR_rank'] = cleaned_data.groupby('year')['PSR'].transform(lambda x: x.rank(ascending=True))

# Calculate the factor_value
columns_for_factor_value = ['PBR_rank', 'PCR_rank', 'PER_rank', 'PSR_rank']
cleaned_data['factor_value'] = cleaned_data.groupby('year')[columns_for_factor_value].transform(lambda x: (x - np.nanmean(x)) / np.nanstd(x)).sum(axis=1)

# Display the result
cleaned_data[['year', 'PBR_rank', 'PCR_rank', 'PER_rank', 'PSR_rank', 'factor_value']].head()
cleaned_data
```

#### 퀄리티팩터


다음으로 퀄리티 지표에 해당하는 투하자본 수익률을 계산하겠습니다. 해당 값은 이자 및 법인세 차감전이익(EBIT)를 투하자본(IC)으로 나누어 계산되며, 이를 분해하면 다음과 같습니다.

\begin{equation*}
\begin{split}
투하자본\,수익률 & = \frac{이자\,및\,법인세\,차감전이익}{투하자본} \\
& = \frac{당기순이익 + 법인세 + 이자비용}{(유동자산 - 유동부채) + (비유동자산 - 감가상각비)}
\end{split}
\end{equation*}

**Quality index**: 자기자본이익률 roe, 매출총이익 gpa, 영업활동현금흐름 cfo

```python
# Calculate the metrics
cleaned_data['ROE'] = cleaned_data['지배주주순이익(천원)'] / cleaned_data['총자본(천원)']
cleaned_data['GPA'] = cleaned_data['매출총이익(천원)'] / cleaned_data['총자산(천원)']
cleaned_data['CFO'] = cleaned_data['영업활동으로인한현금흐름(천원)'] / cleaned_data['총자산(천원)']

# Calculate the rank for each metric
cleaned_data['ROE_rank'] = cleaned_data.groupby('year')['ROE'].transform(lambda x: x.rank(ascending=True))
cleaned_data['GPA_rank'] = cleaned_data.groupby('year')['GPA'].transform(lambda x: x.rank(ascending=True))
cleaned_data['CFO_rank'] = cleaned_data.groupby('year')['CFO'].transform(lambda x: x.rank(ascending=True))

# Calculate the factor_quality
columns_for_factor_quality = ['ROE_rank', 'GPA_rank', 'CFO_rank']
cleaned_data['factor_quality'] = cleaned_data.groupby('year')[columns_for_factor_quality].transform(lambda x: (x - np.nanmean(x)) / np.nanstd(x)).sum(axis=1)

# Display the result
cleaned_data[['year', 'ROE_rank', 'GPA_rank', 'CFO_rank', 'factor_quality']].head()
```

##모멘텀팩터

```python
#Momentum factor_근 3개월, 최근 6개월, 최근 12개월 기간 동안의 수익률, 단, 최근 1개월은 수익률 계산하는 샘플 기간에서 뺌

# Group by Symbol and year, then apply the momentum calculations
def compute_momentum_factors(group):
    # Calculate the logarithmic returns
    ret = np.log(group['종가(원)']) - np.log(group['종가(원)']).shift(1)

    # Compute the momentum factors
    mom_3m = (1 + ret).iloc[-60:].prod() - 1
    mom_6m = (1 + ret).iloc[-120:].prod() - 1
    mom_12m = (1 + ret).iloc[-252:].prod() - 1

    return pd.Series({'mom_3m': mom_3m, 'mom_6m': mom_6m, 'mom_12m': mom_12m})

# Grouping by stock Symbol and year, and applying the function
momentum_data = full_stock_df.groupby(['code', 'year']).apply(compute_momentum_factors).reset_index()

# Ranking the momentum factors in descending order for each year
momentum_data[['mom_3m_rank', 'mom_6m_rank', 'mom_12m_rank']] = momentum_data[['year', 'mom_3m', 'mom_6m', 'mom_12m']].groupby('year').rank(ascending=False, method='min')

# Calculating the scaled ranking for each momentum rank and summing across the three momentum factors
momentum_data['factor_mom'] = momentum_data[['year', 'mom_3m_rank', 'mom_6m_rank', 'mom_12m_rank']].groupby('year').transform(lambda x: (x - np.nanmean(x)) / np.nanstd(x)).sum(axis=1)


momentum_data.head()
```

##Growth factor

```python
#Growth factor
# Calculating log growth rates for the specified indices
indices = ["매출액(천원)", "매출총이익(천원)", "영업이익(천원)", "지배주주순이익(천원)"]

for index in indices:
    growth_rate_col_name = f"{index}_growth_rate"
    merged_df_final[growth_rate_col_name] = merged_df_final.groupby('Symbol')[index].apply(lambda x: np.log(x / x.shift(1)))

growth_rate_cols = [f"{index}_growth_rate" for index in indices]
merged_df_final['GrowthFactor'] = merged_df_final[growth_rate_cols].mean(axis=1)
merged_df_final['GrowthFactor'].head()
```

```python
def select_top_30_stocks_weighted(data, year, weights):
    """
    Select the top 30 stocks for the given year based on the weighted combined rank of value, quality, momentum, and growth factors.

    Parameters:
    - data: DataFrame containing the stock data
    - year: Target year for which top 30 stocks are to be selected
    - weights: A list or tuple containing weights for value, quality, momentum, and growth factors respectively

    Returns:
    - DataFrame containing the top 30 stocks and their ranks for the specified year
    """

    # Filtering the data for the target year
    year_data = data[data['year'] == year].copy()  # Using copy() to avoid SettingWithCopyWarning

    # Ranking for each factor
    year_data['Value_rank'] = year_data['factor_value'].rank(ascending=False)
    year_data['Quality_rank'] = year_data['factor_quality'].rank(ascending=False)
    year_data['Momentum_rank'] = year_data['factor_mom'].rank(ascending=False)
    year_data['Growth_rank'] = year_data['GrowthFactor'].rank(ascending=False)

    # Calculate the weighted score
    year_data['weighted_score'] = weights[0] * year_data['Value_rank'] + \
                                  weights[1] * year_data['Quality_rank'] + \
                                  weights[2] * year_data['Momentum_rank'] + \
                                  weights[3] * year_data['Growth_rank']

    # Selecting the top 30 stocks based on the weighted score
    top_30_stocks = year_data.nsmallest(30, 'weighted_score')

    return top_30_stocks[['Symbol', 'Symbol Name']]

equal_weights = (0.25, 0.25, 0.25, 0.25)
select_top_30_stocks_weighted(merged_df_final, 2021, equal_weights)
```

```python
##포트폴리오_30 return 및 volatility 계산
def select_top_30_stocks_weighted(data, year, weights, full_stock_df):
    """
    Select the top 30 stocks for the given year based on the weighted combined rank of value, quality, momentum, and growth factors.
    Calculate the 수익률 and volatility for each of the top 30 stocks over the evaluation period.

    Parameters:
    - data: DataFrame containing the stock data
    - year: Target year for which top 30 stocks are to be selected
    - weights: A list or tuple containing weights for value, quality, momentum, and growth factors respectively
    - full_stock_df: DataFrame containing daily stock returns

    Returns:
    - DataFrame containing the top 30 stocks, their average daily return, and annualized volatility for the evaluation period
    """

    # Filtering the data for the target year
    year_data = data[data['year'] == year].copy()  # Using copy() to avoid SettingWithCopyWarning

    # Ranking for each factor
    year_data['Value_rank'] = year_data['factor_value'].rank(ascending=False)
    year_data['Quality_rank'] = year_data['factor_quality'].rank(ascending=False)
    year_data['Momentum_rank'] = year_data['factor_mom'].rank(ascending=False)
    year_data['Growth_rank'] = year_data['GrowthFactor'].rank(ascending=False)

    # Calculate the weighted score
    year_data['weighted_score'] = weights[0] * year_data['Value_rank'] + \
                                  weights[1] * year_data['Quality_rank'] + \
                                  weights[2] * year_data['Momentum_rank'] + \
                                  weights[3] * year_data['Growth_rank']

    # Selecting the top 30 stocks based on the weighted score
    top_30_stocks = year_data.nsmallest(30, 'weighted_score')
    top_30_symbols = top_30_stocks['Symbol'].tolist()

    # Define the evaluation period based on the given year
    start_date = f"{year + 1}-04-01"
    end_date = f"{year + 2}-03-31"

    # Filter the stock data for the evaluation period and the top 30 stocks
    eval_period_data = full_stock_df[(full_stock_df['date'] >= start_date) &
                                     (full_stock_df['date'] <= end_date) &
                                     (full_stock_df['code'].isin(top_30_symbols))]

    # Calculate the average daily return and the annualized volatility for each stock
    stock_summary = eval_period_data.groupby('code').agg(
        수익률=('수익률(%)', 'mean'),
        volatility=('수익률(%)', lambda x: x.std() * np.sqrt(252))  # 252 trading days in a year
    ).reset_index()

    # Merge the summary data with top 30 stocks
    result = pd.merge(top_30_stocks[['Symbol', 'Symbol Name']], stock_summary, left_on='Symbol', right_on='code', how='left')
    result = result[['Symbol', 'Symbol Name', '수익률', 'volatility']]

    return result

equal_weights = (0.00, 0.50, 0.50, 0.00)
select_top_30_stocks_weighted(merged_df_final, 2001, equal_weights, full_stock_df)
```

```python
def portfolio_stats(data, year, weights, full_stock_df):

    # Filtering the data for the target year
    year_data = data[data['year'] == year].copy()

    # Ranking for each factor
    year_data['Value_rank'] = year_data['factor_value'].rank(ascending=False)
    year_data['Quality_rank'] = year_data['factor_quality'].rank(ascending=False)
    year_data['Momentum_rank'] = year_data['factor_mom'].rank(ascending=False)
    year_data['Growth_rank'] = year_data['GrowthFactor'].rank(ascending=False)

    # Calculate the weighted score
    year_data['weighted_score'] = sum(w * year_data[factor + '_rank'] for w, factor in zip(weights, ['Value', 'Quality', 'Momentum', 'Growth']))

    # Selecting the top 30 stocks based on the weighted score
    top_30_stocks = year_data.nsmallest(30, 'weighted_score')
    top_30_symbols = top_30_stocks['Symbol'].tolist()

    # Define the evaluation period based on the given year
    start_date = f"{year + 1}-04-01"
    end_date = f"{year + 2}-03-31"

    # Filter the stock data for the evaluation period and the top 30 stocks
    eval_period_data = full_stock_df[(full_stock_df['date'] >= start_date) &
                                     (full_stock_df['date'] <= end_date) &
                                     (full_stock_df['code'].isin(top_30_symbols))]

    # Calculate the average daily return and the annualized volatility for each stock
    stock_summary = eval_period_data.groupby('code').agg(
        수익률=('수익률(%)', 'mean'),
        volatility=('수익률(%)', lambda x: x.std() * np.sqrt(252))  # 252 trading days in a year
    ).reset_index()

    # Calculate portfolio return and volatility
    portfolio_return = stock_summary['수익률'].mean()
    portfolio_volatility = np.sqrt((stock_summary['volatility']**2).mean())

    return portfolio_return, portfolio_volatility
```

```python
# Create empty dataframes to store results
portfolio_returns_df = pd.DataFrame(index=range(2001, 2021), columns=strategy_weights.keys())
portfolio_volatility_df = pd.DataFrame(index=range(2001, 2021), columns=strategy_weights.keys())


# Loop through each year and each strategy weight
for year in range(2001, 2021):
    for strategy, weight in strategy_weights.items():
        try:
            # Get the top 30 stocks for the given year and weight
            top_30_stocks = select_top_30_stocks_weighted(merged_df_final, year, weight, stock_df)

            # Compute the portfolio return and volatility
            portfolio_return = top_30_stocks['수익률'].mean()
            portfolio_volatility = np.sqrt((top_30_stocks['volatility']**2).mean())

            # Store the results in respective dataframes
            portfolio_returns_df.at[year, strategy] = portfolio_return
            portfolio_volatility_df.at[year, strategy] = portfolio_volatility

        except Exception as e:
            print(f"Error for year {year} and strategy {strategy}: {e}")

portfolio_returns_df, portfolio_volatility_df
```
