```python
import aiohttp
!pip install nest_asyncio

# Import libraries
import nest_asyncio
import asyncio

# Apply nest_asyncio to handle event loop issues
nest_asyncio.apply()
from bs4 import BeautifulSoup
import pandas as pd
from tqdm.asyncio import tqdm_asyncio

# Base URLs
base_url = "https://unibook.unikorea.go.kr/material/view?"
main_list_url = "https://unibook.unikorea.go.kr/material/list?materialScope=NART&configCodes=ART&fields=PUBLISHER&keywords=%EB%A1%9C%EB%8F%99%EC%8B%A0%EB%AC%B8%EC%82%AC&conjunctions=OR&fields=PUBLISHER&keywords=%EB%AF%BC%EC%A3%BC%EC%A1%B0%EC%84%A0%ED%8E%B8%EC%A7%91%EC%9C%84%EC%9B%90%ED%9A%8C&conjunctions=OR&fields=PUBLISHER&keywords=%EB%AC%B8%ED%95%99%EC%8B%A0%EB%AC%B8%EC%82%AC&conjunctions=OR&fields=PUBLISHER&keywords=%ED%8F%89%EC%96%91%ED%83%80%EC%9E%84%EC%A6%88&page="

# Define a function to extract the text using CSS selectors
def get_text_by_xpath(soup, xpath):
    elements = soup.select(xpath)
    if elements:
        return elements[0].get_text(strip=True)
    return None

# Asynchronous function to fetch content
async def fetch(session, url):
    async with session.get(url) as response:
        return await response.text()

# Asynchronous function to process subpages
async def process_subpage(session, subpage_url):
    html_content = await fetch(session, subpage_url)
    sub_soup = BeautifulSoup(html_content, "html.parser")

    # Extract fields using updated CSS selectors
    return {
        "주기사명": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(1) > dd"),
        "서명": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(2) > dd"),
        "권호": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(3) > dd"),
        "발행처": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(4) > dd"),
        "발행일자": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(5) > dd"),
        "페이지": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(6) > dd"),
        "저자": get_text_by_xpath(sub_soup, "div.biblio_info > div.info_list > dl:nth-of-type(7) > dd"),
    }

# Asynchronous function to scrape a single page
async def scrape_page(session, page_number):
    page_url = main_list_url + str(page_number)
    html_content = await fetch(session, page_url)
    soup = BeautifulSoup(html_content, "html.parser")

    # Extract subpage links
    subpage_links = []
    for link in soup.select("h6 > a[href]"):
        subpage_links.append(base_url + link["href"].split("?")[-1])

    # Process each subpage asynchronously
    tasks = [process_subpage(session, subpage_url) for subpage_url in subpage_links]
    return await asyncio.gather(*tasks)

# Main function to scrape multiple pages asynchronously
async def main(start_page, end_page):
    async with aiohttp.ClientSession() as session:
        all_data = []
        for page_number in tqdm_asyncio(range(start_page, end_page + 1), desc="Scraping Pages"):
            all_data.extend(await scrape_page(session, page_number))
        return all_data

# Run the scraper
start_page = 1
end_page = 10000  # Use a smaller range for testing

data = asyncio.run(main(start_page, end_page))

# Convert to DataFrame
df = pd.DataFrame(data)
```
